Confusion Matrix Analysis

The SVM model produced the best confusion matrix results, with only 17 misclassifications out of 20,000 test samples. This indicates excellent classification performance and explains its highest accuracy score of 99.96%.

Random Forest also performed well, correctly classifying most campaigns and making relatively few errors. Decision Tree achieved good performance but produced more false negatives compared to Random Forest and SVM.

KNN showed moderate performance with a higher number of misclassifications, resulting in lower accuracy. Naive Bayes had the highest number of classification errors, which led to the lowest accuracy among all models.

Overall, the confusion matrices confirm that SVM is the most effective model for predicting crowdfunding campaign success, followed by Random Forest and Decision Tree.

SVM

Strengths: Highest accuracy and best overall performance.

Weaknesses: Requires feature scaling and can be slower to train.

Improvements: Tune C and try different kernels.

Random Forest

Strengths: High accuracy and handles complex data well.

Weaknesses: Harder to interpret than a Decision Tree.

Improvements: Tune n_estimators and max_depth.

Decision Tree

Strengths: Simple, easy to understand, and fast.

Weaknesses: Can overfit the data.

Improvements: Adjust tree depth and pruning settings.

KNN

Strengths: Simple and effective after scaling.

Weaknesses: Sensitive to the choice of k.

Improvements: Optimize n_neighbors and distance metrics.

Naive Bayes

Strengths: Fast and easy to implement.

Weaknesses: Assumes features are independent.

Improvements: Tune var_smoothing and improve feature selection.

Conclusion

SVM achieved the best performance and was selected as the final model. Random Forest and Decision Tree also performed well, while KNN showed moderate performance. Naive Bayes had the lowest accuracy.
